# Kenya Household Cash-Flow Analytics

### Understanding Household Financial Behaviour and Cash-Flow Management in Kenya

## 1. Business Understanding

### 1.1 Business Context

Financial inclusion is not simply about providing access to financial services. It is also about ensuring that those services reflect how people earn, spend, save, borrow, and manage their money.

Many low-income households in Kenya rely on multiple sources of income, including casual employment, small businesses, agriculture, formal employment, and financial support from family or social networks. These income sources may be irregular, unpredictable, or seasonal.

Household financial obligations, however, are not always aligned with income receipts. Everyday expenses such as food and transportation must be met continuously, while obligations such as school fees, rent, and healthcare may require substantial payments at particular times.

To manage these differences, households may rely on savings, borrowing, informal savings groups, mobile money, and support from relatives or friends.

**Understanding the relationship between income timing, expenditure obligations, and financial management behaviour is essential to designing financial services that reflect households' actual circumstances.**

The Kenya Financial Diaries study, conducted during 2012–2013, provides detailed historical records of household financial activity. These records offer an opportunity to investigate how participating households managed their finances over time.

### 1.2 Business Problem

Financial inclusion researchers, development organizations, and financial service providers need to understand more than how much households earn or spend. They need to understand **when financial resources become available, when expenditure obligations arise, and how households manage differences between the two.**

However, detailed household transaction records are difficult to interpret without appropriate processing and classification.

The Kenya Financial Diaries dataset captures a wide range of financial activities, including income receipts, income-generation costs, household expenditure, borrowing, savings, transfers, and financial-account transactions.

These activities have different economic meanings and cannot be treated as equivalent financial inflows or outflows.

For example:

* A household may receive a loan without generating additional earned income.
* Money deposited into savings represents a movement of financial resources rather than consumption expenditure.
* Goods purchased on credit create an expenditure obligation without necessarily involving an immediate cash payment.
* A household may experience expenditure exceeding recorded income in a particular period while relying on previously accumulated savings.

Additionally, differences in household observation periods, repeated transaction identifiers, and incomplete records complicate consistent comparisons.

**The central business problem is the difficulty of converting complex household financial transactions into reliable, comparable information that explains how households manage irregular financial resources and expenditure obligations over time.**

Without this information, researchers and programme managers may overlook important differences in household financial behaviour or draw misleading conclusions from aggregated financial figures.

### 1.3 Central Business Question

**How do low-income Kenyan households manage the timing and variability of income and expenditure, and what financial mechanisms do they use during periods when recorded financial resources and spending obligations do not align?**

### 1.4 Project Objective

The objective of this project is to develop a reliable, reproducible analytical foundation for understanding household cash-flow patterns using historical Kenya Financial Diaries data.

The project will organize, validate, and classify household financial transactions to support meaningful analysis of income-generating activities, expenditure patterns, and financial management mechanisms.

The resulting analysis will help researchers understand differences in household financial behaviour and investigate how households respond to irregular income and substantial financial obligations.

### 1.5 Key Business Questions

The project will investigate four interconnected questions.

**1. Income Variability**

How do household income-generating activities vary over time, and how consistent or irregular are recorded income patterns across households?

This will help researchers understand differences in the frequency, timing, and variability of household earnings.

**2. Expenditure Patterns and Financial Obligations**

What are the major household expenditure categories, and how do the frequency, value, and timing of expenses differ?

Particular attention will be given to understanding the distinction between frequent everyday expenses and occasional substantial obligations, such as education and healthcare costs.

**3. Household Financial Management**

How do households use borrowing, savings, transfers, and informal financial arrangements to manage their financial activities?

This will support investigation of the role different financial mechanisms play in household money management.

**4. Potential Cash-Flow Mismatches**

During which periods does recorded operational expenditure exceed defined income-related resources, and what other financial activities occur during those periods?

This will help researchers investigate potential timing mismatches without assuming that negative financial margins necessarily indicate poverty or financial distress.

### 1.6 Intended Stakeholders

The primary stakeholders are **financial inclusion researchers and development programme managers** seeking evidence about how low-income households manage their financial resources.

Other potential users include:

* Financial institutions researching products for customers with irregular income.
* Development finance organizations investigating barriers to financial inclusion.
* NGOs designing household financial resilience and savings programmes.
* Policy researchers studying household financial behaviour.

### 1.7 Expected Business Value

The project aims to make complex historical household financial records more accessible, consistent, and meaningful for financial inclusion research.

By examining the relationship between income variability, spending obligations, and financial management activities, the analysis can help stakeholders:

* **Understand income irregularity:** Investigate how income timing and variability differ across households.
* **Identify expenditure patterns:** Distinguish frequent everyday expenses from less frequent, substantial financial obligations.
* **Understand financial management behaviour:** Examine how savings, borrowing, transfers, and informal financial arrangements feature in household financial activity.
* **Investigate potential financial pressure:** Identify periods when recorded income-related resources and expenditure do not align, while recognizing the limitations of these measures.
* **Inform programme research:** Generate evidence and research questions relevant to flexible savings arrangements, repayment schedules, and financial planning initiatives.

The intended value is not to prescribe specific financial products, but to provide a stronger evidence base for understanding household financial needs and informing further research and programme development.

### 1.8 Scope and Limitations

The analysis will focus on historical household financial activity recorded in the Kenya Financial Diaries study during 2012–2013.

Several considerations are important when interpreting the findings:

* **Historical relevance:** The study provides evidence about household financial behaviour during the observation period, not a direct representation of current financial conditions in Kenya.
* **Sample coverage:** The participating households are not nationally representative. Findings should not be generalized to all Kenyan households.
* **Financial interpretation:** Income, expenditure, borrowing, savings, transfers, and financial-account movements must be distinguished when defining analytical measures.
* **Cash-flow interpretation:** Recorded income and expenditure differences do not necessarily represent actual cash availability or financial distress.
* **Data completeness:** Differences in household observation periods and data quality must be considered when comparing financial activity.
* **Analytical limitations:** Observed relationships between financial activities do not establish causation or demonstrate the effectiveness of particular financial products.

The project will therefore emphasize **descriptive, evidence-based analysis**, supported by clearly documented financial definitions and analytical assumptions.

### 1.9 Project Outcome

The expected outcome is a reliable analytical foundation that enables stakeholders to investigate how participating Kenyan households earned, spent, and managed financial resources over time.

By transforming detailed financial transactions into consistent household-level information, the project will make it easier to explore income variability, expenditure obligations, and the financial mechanisms associated with periods of potential cash-flow mismatch.

**Ultimately, the project seeks to bridge the gap between raw household financial records and actionable research evidence, supporting a better understanding of the financial realities that should inform inclusive financial services and development programmes.**

### Understanding unique_hhs

```python
# Investigate the unique_hhs column

# Select records where unique_hhs has a value
hh_flagged = df[df["unique_hhs"].notna()]

# Compare household counts
print("Total households:", df["hh_ids"].nunique())

print("Rows with unique_hhs populated:", len(hh_flagged))

print("Distinct households in flagged rows:",
      hh_flagged["hh_ids"].nunique())

print("Values in unique_hhs:",
      hh_flagged["unique_hhs"].unique())

print("Households flagged more than once:",
      hh_flagged["hh_ids"].duplicated().sum())
```

This is evidence that unique_hhs acts as a counting indicator marking one record per household.
we will use `hh_ids` as the household identifier.

```python
# Count transaction families within each financial class

family_counts = (
    df.groupby(
        ["trx_class_code", "trx_family_desc"],
        dropna=False
    )
    .size()
    .reset_index(name="transaction_count")
    .sort_values(
        ["trx_class_code", "transaction_count"],
        ascending=[True, False]
    )
)

display(family_counts)
```
